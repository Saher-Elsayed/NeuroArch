# End-Use Energy Breakdown
Reproduces paper Fig. 12

```python
import pandas as pd, matplotlib.pyplot as plt
df = pd.read_csv('../data/energyplus/medium_office/end_use_monthly.csv')
months = df['month'].tolist()
x = range(len(months))
fig, ax = plt.subplots(figsize=(10,4))
ax.bar(x, df['HVAC_Thermal_kWh_m2'], label='HVAC Thermal', color='steelblue')
ax.bar(x, df['Fans_Pumps_kWh_m2'], bottom=df['HVAC_Thermal_kWh_m2'], label='Fans+Pumps', color='skyblue')
stack = df['HVAC_Thermal_kWh_m2']+df['Fans_Pumps_kWh_m2']
ax.bar(x, df['Interior_Lighting_kWh_m2'], bottom=stack, label='Lighting', color='gold')
ax.bar(x, df['Other_kWh_m2'], bottom=stack+df['Interior_Lighting_kWh_m2'], label='Other', color='lightgray')
ax.set_xticks(x); ax.set_xticklabels(months, rotation=30)
ax.set_ylabel('kWh/m²/month'); ax.set_title('NeuroArch Monthly End-Use Energy (Medium Office)')
ax.legend(loc='upper left')
plt.tight_layout(); plt.savefig('energy_enduse.png',dpi=150); plt.show()
print(f'Annual total: {df["Total_kWh_m2"].sum():.1f} kWh/m² (paper: 108.6)')
```
