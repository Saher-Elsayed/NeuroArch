# User Study Statistical Analysis
Reproduces paper Table 5

```python
import pandas as pd, numpy as np
from scipy import stats
tt = pd.read_csv('../data/user_study/task_times.csv')
tlx= pd.read_csv('../data/user_study/nasa_tlx.csv')
sus= pd.read_csv('../data/user_study/sus_scores.csv')
ssq= pd.read_csv('../data/user_study/ssq_subscales.csv')
print('N =', len(tt))

def report(name, base, vr):
    t,p = stats.ttest_rel(base,vr)
    d = (np.mean(base)-np.mean(vr))/np.std(np.array(base)-np.array(vr),ddof=1)
    print(f'{name}: Baseline={np.mean(base):.1f}±{np.std(base):.1f}  VR={np.mean(vr):.1f}±{np.std(vr):.1f}  p={p:.4f}  d={d:.2f}')

report('T1 time (s)',  tt.T1_baseline_s, tt.T1_vr_s)
report('T2 time (s)',  tt.T2_baseline_s, tt.T2_vr_s)
report('NASA-TLX',    tlx.cond_baseline_total, tlx.cond_vr_total)
print(f'SUS mean: {sus.sus_total.mean():.1f} ± {sus.sus_total.std():.1f} (Good = >80.3)')
print(f'SSQ total: {ssq.ssq_total.mean():.1f} | Nausea: {ssq.nausea.mean():.1f} | Oculomotor: {ssq.oculomotor.mean():.1f}')
red = 1-(tt.T1_vr_s.mean()+tt.T2_vr_s.mean())/(tt.T1_baseline_s.mean()+tt.T2_baseline_s.mean())
print(f'Task-time reduction: {red*100:.1f}% (paper: 41.3%)')
```
