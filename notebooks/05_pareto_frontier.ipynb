# Pareto Frontier: Reward Weight Sensitivity
Reproduces paper Table 12

```python
import matplotlib.pyplot as plt, numpy as np
results = [
    (0.0, 104.1, 84.2, '$\\lambda_C=0$ (energy only)'),
    (1.0, 111.3, 89.8, '$\\lambda_C=1$'),
    (2.0, 108.6, 91.3, '$\\lambda_C=2$ (NeuroArch)'),
    (4.0, 115.2, 91.7, '$\\lambda_C=4$'),
    (999, 138.4, 93.1, 'comfort only'),
]
lc = [r[0] for r in results]; e = [r[1] for r in results]; c = [r[2] for r in results]
fig,ax = plt.subplots(figsize=(7,5))
ax.scatter(e,c,s=100,zorder=5)
ax.plot(e,c,'--',alpha=0.4)
for i,(r) in enumerate(results):
    ax.annotate(r[3],(e[i],c[i]),textcoords='offset points',xytext=(5,5),fontsize=9)
ax.axvline(108.6,color='red',linestyle=':',alpha=0.5,label='NeuroArch')
ax.set_xlabel('Annual Energy (kWh/m²)'); ax.set_ylabel('ASHRAE 55 Compliance (%)')
ax.set_title('Energy-Comfort Pareto Frontier')
ax.legend(); plt.tight_layout(); plt.savefig('pareto.png',dpi=150); plt.show()
```
