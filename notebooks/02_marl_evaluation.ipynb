# MARL Policy Evaluation
Reproduces Table 4 energy/comfort results

```python
import csv, pandas as pd
df = pd.read_csv('../data/energyplus/medium_office/controller_comparison.csv')
df['energy_saving_pct'] = df['energy_saving_pct'].astype(float)
print(df.to_string(index=False))
print(f"\nNeuroArch saving: {df[df.controller=='NeuroArch_QMIX']['energy_saving_pct'].values[0]}%")
print("Expected: -23.7% (Table 4)")
```

```python
import matplotlib.pyplot as plt
colors = ['gray','steelblue','green','orange','red']
for i,(_, row) in enumerate(df.iterrows()):
    plt.bar(i, float(row['annual_kWh_m2']), color=colors[i], alpha=0.8, label=row['controller'])
plt.ylabel('Annual Energy Intensity (kWh/m²)')
plt.title('Controller Comparison – Medium Office')
plt.xticks(range(len(df)), [r[:12] for r in df.controller], rotation=20, ha='right')
plt.tight_layout(); plt.savefig('controller_comparison.png',dpi=150); plt.show()
```
